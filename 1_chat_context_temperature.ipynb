```python
# create anthropic client

from anthropic import Anthropic
client = Anthropic()
```

```python
model = "claude-sonnet-4-6"
```

```python
# create first msg request to cleint 



message = client.messages.create(
    model = model,
    max_tokens = 100,
    messages = [
        {
            "role": "user",
            "content": "What is iOS, answer in one sentence"
        }

    ]
)

# try to see if context is preserved  - we saw it dont preserve previouse messages in the next request.

message = client.messages.create(
    model = model,
    max_tokens = 100,
    messages = [
        {
            "role": "user",
            "content": "tell me more about it in one sentence"
        }

    ]
)
```

```python

print(message.content[0].text)
```

```python


def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

def add_assistant_message(messages, text):
    messages.append(
        {
            "role": "assistant",
            "content": text
        }
    )

def chat(messages, system=None, temp=0.1):
    all_params = {
        "model": model,
        "max_tokens": 100,
        "messages": messages,
    

    }

    if system:
        all_params["system"] = system
    
    message = client.messages.create(**all_params)

    return  message.content[0].text
    

```

```python
op = chat(
        [{
        "role" : "user",
        "content" : "give me one drawing idea for cartoon character, keep it brief"
        }]
        )
print(op)
```


### lets create live chat bot using built in input function to practice what we learnt above . 


```python
questions = [] 

system_prompt = '''
    You are a rude assistant that answers questions in a concise manner, mostly in few words.
    '''

while True:
    user_input = input("What is the question")
    print(user_input)

    # add this to list 
    add_user_message(questions, user_input)

    # call chat 
    answer = chat(questions, system_prompt)

    print('Response: ', answer)

    add_assistant_message(questions,answer)

    choice = input('do you want to continue ? (y/n)')

    if choice != 'y':
        print('exisitng chat')
        break
```
